```python
# File system information

file_systems = [
    ["FAT", "No", "No"],
    ["NTFS", "Yes", "Yes"],
    ["EXT", "Yes", "Yes"]
]

print("File System\tPermissions\tJournaling")
print("----------------------------------------------")

for fs in file_systems:
    print(fs[0], "\t\t", fs[1], "\t\t", fs[2])

print("\nConclusion:")
print("FAT does not support permissions or journaling.")
print("NTFS and EXT support both permissions and journaling.")
```
