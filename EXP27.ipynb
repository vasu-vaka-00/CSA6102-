```python
import hashlib
import shutil

# Step 1: Acquire (Create original evidence)
with open("evidence.txt", "w") as f:
    f.write("Digital Forensics Evidence")

# Step 2: Create a copy
shutil.copy("evidence.txt", "evidence_copy.txt")

# Function to calculate SHA-256 hash
def get_hash(file):
    with open(file, "rb") as f:
        return hashlib.sha256(f.read()).hexdigest()

# Step 3: Hash and Verify
original_hash = get_hash("evidence.txt")
copy_hash = get_hash("evidence_copy.txt")

print("Original Hash:", original_hash)
print("Copy Hash    :", copy_hash)

if original_hash == copy_hash:
    print("\nVerification: PASS")
else:
    print("\nVerification: FAIL")

# Step 4: Tamper with the copy
with open("evidence_copy.txt", "a") as f:
    f.write("\nTampered Data")

tampered_hash = get_hash("evidence_copy.txt")

print("\nAfter Tampering")
print("Tampered Hash:", tampered_hash)

# Step 5: Analyze and Report
if original_hash == tampered_hash:
    print("Integrity Verified")
else:
    print("Integrity Failed")
    print("Report: Evidence has been modified (tampered).")
```
