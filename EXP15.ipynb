```python
# ==========================================================
# TITLE: Volatile vs Non-Volatile Digital Evidence Simulation
# Description:
# Demonstrates volatile evidence (RAM) and non-volatile
# evidence (disk), and verifies integrity using MD5 hashing.
# Runs directly in Google Colab.
# ==========================================================

import hashlib
import time

print("="*70)
print("      VOLATILE vs NON-VOLATILE DIGITAL EVIDENCE")
print("="*70)

# ----------------------------------------------------------
# Volatile Evidence (Stored in Memory)
# ----------------------------------------------------------
print("\n1. Collecting Volatile Evidence (Running Processes)\n")

running_processes = [
    "chrome.exe",
    "python.exe",
    "explorer.exe",
    "cmd.exe",
    "notepad.exe"
]

print("Running Processes:")
for process in running_processes:
    print(" -", process)

# ----------------------------------------------------------
# Non-Volatile Evidence (Stored on Disk)
# ----------------------------------------------------------
print("\n2. Creating Non-Volatile Evidence File\n")

filename = "Evidence_Report.txt"

content = """
Digital Forensic Investigation
Case ID : DF-2026-001
Examiner : Investigator
Status : Evidence Collected Successfully
"""

with open(filename, "w") as file:
    file.write(content)

print("File Created:", filename)

# ----------------------------------------------------------
# Calculate Original MD5 Hash
# ----------------------------------------------------------
with open(filename, "rb") as file:
    data = file.read()

original_hash = hashlib.md5(data).hexdigest()

print("\nOriginal MD5 Hash")
print(original_hash)

# ----------------------------------------------------------
# Simulate System Shutdown
# ----------------------------------------------------------
print("\n3. Simulating System Shutdown...\n")
time.sleep(2)

# Volatile evidence disappears
running_processes = None

print("Volatile Evidence after Shutdown:")
print(running_processes)

# ----------------------------------------------------------
# Simulate Restart and Read File Again
# ----------------------------------------------------------
print("\n4. Reading Non-Volatile Evidence After Restart\n")

with open(filename, "rb") as file:
    new_data = file.read()

new_hash = hashlib.md5(new_data).hexdigest()

print("New MD5 Hash")
print(new_hash)

# ----------------------------------------------------------
# Integrity Verification
# ----------------------------------------------------------
print("\n5. Integrity Verification\n")

if original_hash == new_hash:
    print("PASS: Non-volatile evidence is unchanged.")
    print("The file persisted after shutdown.")
else:
    print("FAIL: Evidence was modified!")

print("\n" + "="*70)
print("SUMMARY")
print("="*70)
print("Volatile Evidence    : Running Processes")
print("Status After Shutdown: Lost")
print()
print("Non-Volatile Evidence: Evidence_Report.txt")
print("Status After Shutdown: Preserved")
print("MD5 Verified         :", original_hash == new_hash)
print("="*70)
```
