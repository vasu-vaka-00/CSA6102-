```python
import os
import hashlib
import shutil

# Step 1: Create a simulated storage device (binary file)
original_file = "simulated_storage_device.img"

with open(original_file, "wb") as f:
    f.write(os.urandom(5 * 1024 * 1024))   # Create a 5 MB binary file

print("Original storage device created.")

# Step 2: Create a bit-stream copy
copy_file = "bitstream_copy.img"
shutil.copyfile(original_file, copy_file)

print("Bit-stream copy created.")

# Step 3: Compare file sizes
original_size = os.path.getsize(original_file)
copy_size = os.path.getsize(copy_file)

print("\nFile Size Comparison")
print("---------------------")
print("Original Size :", original_size, "bytes")
print("Copy Size     :", copy_size, "bytes")

if original_size == copy_size:
    print("✓ Sizes Match")
else:
    print("✗ Sizes Do Not Match")

# Step 4: Function to calculate SHA-256 hash
def sha256_hash(filename):
    sha256 = hashlib.sha256()
    with open(filename, "rb") as f:
        while True:
            data = f.read(4096)
            if not data:
                break
            sha256.update(data)
    return sha256.hexdigest()

# Step 5: Calculate hashes
original_hash = sha256_hash(original_file)
copy_hash = sha256_hash(copy_file)

print("\nSHA-256 Hash Comparison")
print("------------------------")
print("Original Hash :", original_hash)
print("Copy Hash     :", copy_hash)

if original_hash == copy_hash:
    print("✓ Hashes Match")
    print("\nResult: The bit-stream copy is a forensically exact replica of the original storage device.")
else:
    print("✗ Hashes Do Not Match")
    print("\nResult: The copy is NOT an exact forensic replica.")
```
