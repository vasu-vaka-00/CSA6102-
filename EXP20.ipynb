```python
import hashlib

print("=" * 60)
print("HASHING ALGORITHMS - AVALANCHE EFFECT DEMONSTRATION")
print("=" * 60)

# Original and modified text
text1 = "hello"
text2 = "Hello"

# Function to generate hashes
def generate_hashes(text):
    md5 = hashlib.md5(text.encode()).hexdigest()
    sha1 = hashlib.sha1(text.encode()).hexdigest()
    sha256 = hashlib.sha256(text.encode()).hexdigest()
    return md5, sha1, sha256

# Generate hashes
md5_1, sha1_1, sha256_1 = generate_hashes(text1)
md5_2, sha1_2, sha256_2 = generate_hashes(text2)

# Display hashes
print("\nOriginal Text :", text1)
print("MD5    :", md5_1)
print("SHA-1  :", sha1_1)
print("SHA-256:", sha256_1)

print("\nModified Text :", text2)
print("MD5    :", md5_2)
print("SHA-1  :", sha1_2)
print("SHA-256:", sha256_2)

# Demonstrate Avalanche Effect
print("\n" + "=" * 60)
print("AVALANCHE EFFECT")
print("=" * 60)

print("\nOnly one character changed:")
print(f'"{text1}"  →  "{text2}"')

print("\nHash Comparison:")

print("MD5 Match    :", md5_1 == md5_2)
print("SHA-1 Match  :", sha1_1 == sha1_2)
print("SHA-256 Match:", sha256_1 == sha256_2)

print("\nObservation:")
print("✓ A single-character change produces completely different hashes.")
print("✓ This is called the Avalanche Effect.")
print("✓ Cryptographic hash functions are highly sensitive to input changes.")
```
