```python
import hashlib
import shutil
import os

print("=" * 60)
print("BIT-STREAM FORENSIC COPY SIMULATION")
print("=" * 60)

# -------------------------------------------------
# Step 1: Create a simulated storage device
# -------------------------------------------------
with open("storage_device.bin", "wb") as f:
    f.write(os.urandom(4096))   # 4 KB simulated storage device

print("\nOriginal storage device created.")

# -------------------------------------------------
# Step 2: Create bit-stream (forensic) copy
# -------------------------------------------------
shutil.copyfile("storage_device.bin", "forensic_image.bin")

print("Bit-stream forensic copy created.")

# -------------------------------------------------
# Step 3: Function to calculate SHA-256 hash
# -------------------------------------------------
def sha256(filename):
    h = hashlib.sha256()
    with open(filename, "rb") as f:
        while True:
            data = f.read(4096)
            if not data:
                break
            h.update(data)
    return h.hexdigest()

# -------------------------------------------------
# Step 4: Compare file sizes
# -------------------------------------------------
original_size = os.path.getsize("storage_device.bin")
copy_size = os.path.getsize("forensic_image.bin")

# -------------------------------------------------
# Step 5: Compare SHA-256 hashes
# -------------------------------------------------
original_hash = sha256("storage_device.bin")
copy_hash = sha256("forensic_image.bin")

# -------------------------------------------------
# Step 6: Display Results
# -------------------------------------------------
print("\n" + "=" * 60)
print("FORENSIC VERIFICATION RESULTS")
print("=" * 60)

print(f"Original Size : {original_size} bytes")
print(f"Copy Size     : {copy_size} bytes")

print("\nOriginal SHA-256 Hash:")
print(original_hash)

print("\nCopy SHA-256 Hash:")
print(copy_hash)

print("\nVerification:")

if original_size == copy_size:
    print("✓ File sizes are identical.")
else:
    print("✗ File sizes do not match.")

if original_hash == copy_hash:
    print("✓ SHA-256 hashes match.")
    print("✓ The forensic image is an EXACT bit-stream replica.")
else:
    print("✗ SHA-256 hashes do not match.")
    print("✗ The copy is NOT forensically identical.")
```
