```python
# ============================================================
# Title: Port Scanning Detection using Simulated Network Traffic
# ============================================================

from collections import defaultdict

# Simulated Network Traffic Log
traffic_log = [
    ("10:00:01", "192.168.1.10", "192.168.1.100", 80),
    ("10:00:02", "192.168.1.10", "192.168.1.100", 443),
    ("10:00:03", "192.168.1.10", "192.168.1.100", 21),
    ("10:00:04", "192.168.1.10", "192.168.1.100", 22),
    ("10:00:05", "192.168.1.10", "192.168.1.100", 23),
    ("10:00:06", "192.168.1.10", "192.168.1.100", 25),
    ("10:00:07", "192.168.1.10", "192.168.1.100", 53),
    ("10:00:08", "192.168.1.10", "192.168.1.100", 110),
    ("10:00:09", "192.168.1.10", "192.168.1.100", 135),

    ("10:01:01", "192.168.1.20", "192.168.1.100", 80),
    ("10:01:02", "192.168.1.20", "192.168.1.100", 80),
    ("10:01:03", "192.168.1.20", "192.168.1.100", 443),

    ("10:02:01", "192.168.1.30", "192.168.1.100", 21),
    ("10:02:02", "192.168.1.30", "192.168.1.100", 22),
    ("10:02:03", "192.168.1.30", "192.168.1.100", 23),
    ("10:02:04", "192.168.1.30", "192.168.1.100", 25),
    ("10:02:05", "192.168.1.30", "192.168.1.100", 53),
    ("10:02:06", "192.168.1.30", "192.168.1.100", 80),
    ("10:02:07", "192.168.1.30", "192.168.1.100", 110),
]

# Threshold for detecting a port scan
PORT_THRESHOLD = 6

# Store unique destination ports contacted by each source IP
port_activity = defaultdict(set)

for time, src_ip, dst_ip, port in traffic_log:
    port_activity[src_ip].add(port)

print("=" * 60)
print("PORT SCANNING DETECTION REPORT")
print("=" * 60)

scanner_found = False

for src_ip, ports in port_activity.items():
    print(f"\nSource IP : {src_ip}")
    print("Ports Accessed :", sorted(ports))
    print("Total Unique Ports :", len(ports))

    if len(ports) >= PORT_THRESHOLD:
        print("Status : ALERT - Possible Port Scanning Detected")
        scanner_found = True
    else:
        print("Status : Normal Activity")

if not scanner_found:
    print("\nNo port scanning behavior detected.")

print("\n" + "=" * 60)
print("Analysis Complete")
print("=" * 60)
```
