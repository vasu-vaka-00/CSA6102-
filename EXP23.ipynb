```python
import hashlib

# Function to compute hashes
def compute_hashes(text):
    return {
        "MD5": hashlib.md5(text.encode()).hexdigest(),
        "SHA-1": hashlib.sha1(text.encode()).hexdigest(),
        "SHA-256": hashlib.sha256(text.encode()).hexdigest()
    }

# Original and modified text
text1 = "hello"
text2 = "Hello"

# Compute hashes
hashes1 = compute_hashes(text1)
hashes2 = compute_hashes(text2)

# Display results
print("Original Text :", text1)
print("Modified Text :", text2)
print("=" * 80)

for algo in hashes1:
    print(f"{algo}")
    print("-" * 80)
    print(f"{text1} : {hashes1[algo]}")
    print(f"{text2} : {hashes2[algo]}")
    print("Hashes Match :", hashes1[algo] == hashes2[algo])
    print()

print("=" * 80)
print("Result:")
print("A single-character change ('hello' → 'Hello') produces completely different")
print("MD5, SHA-1, and SHA-256 digests, demonstrating the Avalanche Effect.")
```
