```python
# Calculate Slack Space

file_size = int(input("Enter file size (bytes): "))
cluster_size = int(input("Enter cluster size (bytes): "))

remainder = file_size % cluster_size

if remainder == 0:
    slack_space = 0
else:
    slack_space = cluster_size - remainder

print("\nFile Size      :", file_size, "bytes")
print("Cluster Size   :", cluster_size, "bytes")
print("Slack Space    :", slack_space, "bytes")

print("\nConclusion:")
print("Slack space is the unused space in the last allocated disk cluster.")
```
