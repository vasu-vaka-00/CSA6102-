```python
# ==========================================================
# TITLE: Digital Evidence Collection and Catalog Simulation
# Description:
# Simulates collecting digital evidence and records
# evidence name, type, size, MD5 hash, and acquisition time.
# Runs directly in Google Colab.
# ==========================================================

import hashlib
import os
from datetime import datetime

print("=" * 75)
print("        DIGITAL FORENSIC EVIDENCE INTAKE CATALOG")
print("=" * 75)

# ----------------------------------------------------------
# Simulated Digital Evidence
# ----------------------------------------------------------

evidence = {
    "emails.txt": """
From: alice@example.com
To: bob@example.com
Subject: Project Meeting

Hi Bob,
Please review the confidential project documents before tomorrow's meeting.
Regards,
Alice
""",

    "browser_history.txt": """
09:10 https://www.google.com
09:15 https://github.com
09:30 https://bank.example.com
09:45 https://chat.openai.com
10:05 https://mail.example.com
""",

    "chat_logs.txt": """
John: Hello!
Mary: Did you send the files?
John: Yes, I uploaded them yesterday.
Mary: Great, thank you.
""",

    "system_logs.txt": """
[INFO] User Login Successful
[INFO] USB Device Connected
[WARNING] Multiple Failed Login Attempts
[INFO] Antivirus Scan Completed
"""
}

catalog = []

# ----------------------------------------------------------
# Create Evidence Files
# ----------------------------------------------------------

for filename, content in evidence.items():
    with open(filename, "w", encoding="utf-8") as file:
        file.write(content.strip())

# ----------------------------------------------------------
# Process Evidence
# ----------------------------------------------------------

for filename in evidence.keys():

    # Read file as bytes
    with open(filename, "rb") as file:
        data = file.read()

    # Calculate MD5 Hash
    md5_hash = hashlib.md5(data).hexdigest()

    # File Size
    size = os.path.getsize(filename)

    # Evidence Type
    if "email" in filename:
        evidence_type = "Email"
    elif "browser" in filename:
        evidence_type = "Browser History"
    elif "chat" in filename:
        evidence_type = "Chat Logs"
    elif "system" in filename:
        evidence_type = "System Logs"
    else:
        evidence_type = "Unknown"

    catalog.append([
        filename,
        evidence_type,
        size,
        md5_hash,
        datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    ])

# ----------------------------------------------------------
# Display Catalog
# ----------------------------------------------------------

print("\nEvidence Intake Catalog\n")

print("{:<22} {:<20} {:<10} {:<34} {}".format(
    "Evidence Item",
    "Type",
    "Size(Bytes)",
    "MD5 Hash",
    "Collected Time"))

print("-" * 120)

for item in catalog:
    print("{:<22} {:<20} {:<10} {:<34} {}".format(
        item[0],
        item[1],
        item[2],
        item[3],
        item[4]
    ))

print("\n" + "=" * 75)
print("Total Evidence Items Collected :", len(catalog))
print("=" * 75)

print("\nEvidence successfully cataloged following forensic intake procedures.")
```
