```python
# ==========================================================
# FILE CARVING SIMULATION (JPEG Recovery)
# Google Colab Ready - No Input Required
# ==========================================================

import os

print("=" * 60)
print("FILE CARVING SIMULATION")
print("=" * 60)

# JPEG Start and End Signatures
JPEG_START = b'\xFF\xD8\xFF'
JPEG_END = b'\xFF\xD9'

# ----------------------------------------------------------
# Create a simulated damaged storage device
# ----------------------------------------------------------

raw_data = (
    os.urandom(40) +
    JPEG_START +
    b"Deleted_JPEG_Image_1_Data" +
    JPEG_END +
    os.urandom(50) +
    JPEG_START +
    b"Deleted_JPEG_Image_2_Data" +
    JPEG_END +
    os.urandom(30)
)

with open("damaged_storage.bin", "wb") as f:
    f.write(raw_data)

print("\nSimulated damaged storage created.")

# ----------------------------------------------------------
# File Carving
# ----------------------------------------------------------

with open("damaged_storage.bin", "rb") as f:
    data = f.read()

count = 0
position = 0

while True:
    start = data.find(JPEG_START, position)

    if start == -1:
        break

    end = data.find(JPEG_END, start)

    if end == -1:
        break

    end += len(JPEG_END)

    carved = data[start:end]

    filename = f"Recovered_Image_{count+1}.jpg"

    with open(filename, "wb") as out:
        out.write(carved)

    print(f"\nRecovered: {filename}")
    print(f"Start Offset : {start}")
    print(f"End Offset   : {end-1}")
    print(f"Size         : {len(carved)} bytes")

    count += 1
    position = end

# ----------------------------------------------------------
# Result
# ----------------------------------------------------------

print("\n" + "=" * 60)

if count > 0:
    print(f"Total JPEG Files Recovered : {count}")
else:
    print("No JPEG files found.")

print("=" * 60)

print("\nObservation:")
print("✓ File carving ignores file system records.")
print("✓ It searches raw bytes for JPEG signatures (FFD8FF ... FFD9).")
print("✓ Deleted files can be recovered if their data still exists.")
```
