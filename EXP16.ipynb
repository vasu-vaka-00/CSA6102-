```python
# ==========================================================
# TITLE: Digital Forensic Data Acquisition Types Simulation
# Description:
# Demonstrates Physical, Logical, and Sparse Acquisition
# Runs directly in Google Colab
# ==========================================================

print("=" * 75)
print("      DIGITAL FORENSIC DATA ACQUISITION SIMULATION")
print("=" * 75)

# ----------------------------------------------------------
# Simulated Storage
# ----------------------------------------------------------

storage = [
    {"name": "report.docx", "status": "Visible"},
    {"name": "photo.jpg", "status": "Visible"},
    {"name": "emails.pst", "status": "Visible"},
    {"name": "browser_history.db", "status": "Visible"},
    {"name": "deleted_notes.txt", "status": "Deleted"},
    {"name": "deleted_image.png", "status": "Deleted"},
    {"name": "unused_disk_space", "status": "Unused"}
]

selected_files = ["emails.pst", "browser_history.db"]

# ----------------------------------------------------------
# Physical Acquisition
# ----------------------------------------------------------

print("\n1. PHYSICAL ACQUISITION")
print("-" * 75)

physical = storage

for item in physical:
    print(f"{item['name']:<25} {item['status']}")

print("\nItems Captured:", len(physical))
print("Includes: Visible + Deleted + Unused Disk Space")

# ----------------------------------------------------------
# Logical Acquisition
# ----------------------------------------------------------

print("\n2. LOGICAL ACQUISITION")
print("-" * 75)

logical = [item for item in storage if item["status"] == "Visible"]

for item in logical:
    print(f"{item['name']:<25} {item['status']}")

print("\nItems Captured:", len(logical))
print("Includes: Visible Files Only")

# ----------------------------------------------------------
# Sparse Acquisition
# ----------------------------------------------------------

print("\n3. SPARSE ACQUISITION")
print("-" * 75)

sparse = [item for item in storage if item["name"] in selected_files]

for item in sparse:
    print(f"{item['name']:<25} {item['status']}")

print("\nItems Captured:", len(sparse))
print("Includes: Investigator Selected Files Only")

# ----------------------------------------------------------
# Summary
# ----------------------------------------------------------

print("\n" + "=" * 75)
print("SUMMARY OF ACQUISITION TYPES")
print("=" * 75)

print("{:<22} {:<15} {}".format("Acquisition", "Items", "Captured Data"))
print("-" * 75)

print("{:<22} {:<15} {}".format(
    "Physical",
    len(physical),
    "Visible + Deleted + Unused"
))

print("{:<22} {:<15} {}".format(
    "Logical",
    len(logical),
    "Visible Files Only"
))

print("{:<22} {:<15} {}".format(
    "Sparse",
    len(sparse),
    "Selected Files Only"
))

print("=" * 75)

print("\nKEY DIFFERENCES")
print("✔ Physical Acquisition captures the complete storage image,")
print("  including deleted files and unused disk space.")
print()
print("✔ Logical Acquisition captures only files visible")
print("  through the operating system.")
print()
print("✔ Sparse Acquisition captures only investigator-selected")
print("  files to reduce acquisition time and storage requirements.")
```
