```python
import psutil
import datetime

print("=" * 60)
print(" DIGITAL FORENSICS: LIVE vs DEAD ACQUISITION DEMO")
print("=" * 60)

# -----------------------------
# LIVE ACQUISITION
# -----------------------------
print("\n[1] LIVE ACQUISITION (System ON)")
print("Capturing running processes (volatile data)...\n")

processes = []

for p in psutil.process_iter(['pid', 'name']):
    try:
        processes.append(f"PID: {p.info['pid']}   Process: {p.info['name']}")
    except:
        pass

# Display first 10 processes
for proc in processes[:10]:
    print(proc)

# Save snapshot
with open("process_snapshot.txt", "w") as f:
    f.write("Live Acquisition Snapshot\n")
    f.write("Captured on: " + str(datetime.datetime.now()) + "\n\n")
    for proc in processes:
        f.write(proc + "\n")

print("\nLive process snapshot saved as process_snapshot.txt")

# -----------------------------
# DEAD ACQUISITION
# -----------------------------
print("\n" + "=" * 60)
print("[2] DEAD ACQUISITION (System OFF)")
print("Reading previously saved snapshot...\n")

with open("process_snapshot.txt", "r") as f:
    lines = f.readlines()

for line in lines[:12]:
    print(line.strip())

print("\nDead acquisition can ONLY read the saved snapshot.")
print("It CANNOT recover RAM contents or current running processes.")

# -----------------------------
# COMPARISON
# -----------------------------
print("\n" + "=" * 60)
print("FORENSIC COMPARISON")
print("=" * 60)

print("""
LIVE ACQUISITION
----------------
✓ Captures running processes
✓ Captures volatile memory information
✓ System must be ON

DEAD ACQUISITION
----------------
✓ Reads saved files only
✗ Cannot capture RAM
✗ Cannot recover running processes
✓ Performed after shutdown
""")
```
