```python
import os

# -------------------------------------------------
# Step 1: Create two simulated JPEG files
# -------------------------------------------------
jpeg1 = bytes.fromhex("FFD8FFE000104A4649460001") + os.urandom(200) + bytes.fromhex("FFD9")
jpeg2 = bytes.fromhex("FFD8FFE10018457869660000") + os.urandom(300) + bytes.fromhex("FFD9")

# -------------------------------------------------
# Step 2: Create a damaged raw disk image
# -------------------------------------------------
raw_file = "damaged_disk.img"

with open(raw_file, "wb") as f:
    f.write(os.urandom(500))      # Random bytes
    f.write(jpeg1)                # Hidden JPEG 1
    f.write(os.urandom(800))      # Random bytes
    f.write(jpeg2)                # Hidden JPEG 2
    f.write(os.urandom(400))      # Random bytes

print("Damaged disk image created.")

# -------------------------------------------------
# Step 3: Read the raw disk image
# -------------------------------------------------
with open(raw_file, "rb") as f:
    data = f.read()

# JPEG signatures
start_sig = b'\xFF\xD8\xFF'
end_sig = b'\xFF\xD9'

# -------------------------------------------------
# Step 4: File Carving
# -------------------------------------------------
count = 0
index = 0

while True:
    start = data.find(start_sig, index)

    if start == -1:
        break

    end = data.find(end_sig, start)

    if end == -1:
        break

    end += len(end_sig)

    recovered = data[start:end]

    filename = f"recovered_{count+1}.jpg"

    with open(filename, "wb") as out:
        out.write(recovered)

    print(f"{filename} recovered")
    print(f"Start Offset : {start}")
    print(f"End Offset   : {end}")
    print(f"File Size    : {len(recovered)} bytes\n")

    count += 1
    index = end

print("-----------------------------------")
print("Total JPEG files recovered:", count)
```
