```python
# ===============================================
# FORENSIC IMAGING vs ORDINARY DUPLICATION
# Runs directly in Google Colab
# ===============================================

print("=" * 60)
print("FORENSIC IMAGING vs ORDINARY DUPLICATION")
print("=" * 60)

# Simulated storage sectors
storage = [
    "Boot Record",
    "File A",
    "File B",
    "[Deleted File X]",
    "File C",
    "[Deleted File Y]",
    "Empty Sector",
    "File D",
    "Empty Sector",
    "File E"
]

print("\nOriginal Storage Device (Sector by Sector):\n")

for i, sector in enumerate(storage):
    print(f"Sector {i:02d}: {sector}")

# -----------------------------------------
# Forensic Imaging
# -----------------------------------------
print("\n" + "=" * 60)
print("FORENSIC IMAGING")
print("=" * 60)

forensic_image = storage.copy()

print("\nAll sectors copied (including deleted and empty sectors).\n")

for i, sector in enumerate(forensic_image):
    print(f"Sector {i:02d}: {sector}")

# -----------------------------------------
# Ordinary Duplication
# -----------------------------------------
print("\n" + "=" * 60)
print("ORDINARY FILE DUPLICATION")
print("=" * 60)

ordinary_copy = []

for sector in storage:
    if "Deleted" not in sector and "Empty" not in sector:
        ordinary_copy.append(sector)

print("\nOnly active files copied.\n")

for i, data in enumerate(ordinary_copy):
    print(f"Copied Item {i+1}: {data}")

# -----------------------------------------
# Comparison
# -----------------------------------------
print("\n" + "=" * 60)
print("FORENSIC COMPARISON")
print("=" * 60)

print(f"""
Total Sectors in Original Device : {len(storage)}
Sectors in Forensic Image        : {len(forensic_image)}
Files in Ordinary Copy           : {len(ordinary_copy)}

Observation:
-------------
✓ Forensic Imaging copies every sector.
✓ Deleted files are preserved.
✓ Empty sectors are preserved.
✓ Ordinary duplication copies only visible files.
✓ Deleted data is NOT copied.
""")
```
