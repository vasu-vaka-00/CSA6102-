```python
# ============================================================
# Title: Display Image Metadata (No User Input Required)
# ============================================================

from PIL import Image, PngImagePlugin

# Create a sample image
img = Image.new("RGB", (300, 200), color="lightblue")

# Add metadata
metadata = PngImagePlugin.PngInfo()
metadata.add_text("Author", "Harsha")
metadata.add_text("Description", "Sample Forensic Image")
metadata.add_text("Created", "25-07-2026")
metadata.add_text("Software", "Python Pillow")

# Save image with metadata
img.save("sample_image.png", pnginfo=metadata)

# Open image
image = Image.open("sample_image.png")

print("=" * 50)
print("IMAGE METADATA")
print("=" * 50)

for key, value in image.info.items():
    print(f"{key:15}: {value}")

print("=" * 50)
```
