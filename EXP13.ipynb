```python
# ============================================
# Digital Forensic Report Generator
# Runs directly in Google Colab
# No input required
# ============================================

from datetime import datetime

print("=" * 70)
print("          DIGITAL FORENSIC INVESTIGATION REPORT")
print("=" * 70)

# Sample Investigation Details
case_id = "DF-2026-101"
examiner = "Harsha Vardhan"
organization = "Cyber Security Investigation Lab"
date = datetime.now().strftime("%d-%m-%Y %H:%M:%S")

# Investigation Findings
findings = {
    "Case Type": "Unauthorized System Access",
    "Device Examined": "Windows 11 Laptop",
    "Operating System": "Windows 11 Pro",
    "Evidence Collected": [
        "System Event Logs",
        "Browser History",
        "Deleted Files",
        "USB Device History",
        "Network Logs"
    ],
    "Recovered Files": 18,
    "Suspicious Files": [
        "keylogger.exe",
        "hidden_script.ps1",
        "passwords.txt"
    ],
    "Failed Login Attempts": 37,
    "Malware Detected": "Trojan.Generic",
    "USB Devices Connected": 4,
    "Network Connections": 52,
    "Risk Level": "High"
}

# Report Sections
print("\n1. CASE INFORMATION")
print("-" * 50)
print("Case ID           :", case_id)
print("Examiner          :", examiner)
print("Organization      :", organization)
print("Date & Time       :", date)

print("\n2. INVESTIGATION SUMMARY")
print("-" * 50)
print("Case Type         :", findings["Case Type"])
print("Device Examined   :", findings["Device Examined"])
print("Operating System  :", findings["Operating System"])

print("\n3. EVIDENCE COLLECTED")
print("-" * 50)
for item in findings["Evidence Collected"]:
    print("•", item)

print("\n4. ANALYSIS RESULTS")
print("-" * 50)
print("Recovered Files        :", findings["Recovered Files"])
print("Failed Login Attempts  :", findings["Failed Login Attempts"])
print("Malware Detected       :", findings["Malware Detected"])
print("USB Devices Connected  :", findings["USB Devices Connected"])
print("Network Connections    :", findings["Network Connections"])

print("\nSuspicious Files:")
for file in findings["Suspicious Files"]:
    print(" -", file)

print("\n5. CONCLUSION")
print("-" * 50)

if findings["Risk Level"] == "High":
    print("The investigation indicates evidence of unauthorized access.")
    print("Malware was detected and multiple suspicious files were found.")
    print("The system is considered COMPROMISED.")
else:
    print("No significant malicious activity detected.")

print("\n6. RECOMMENDATIONS")
print("-" * 50)
recommendations = [
    "Isolate the affected device from the network.",
    "Remove detected malware.",
    "Reset all user passwords.",
    "Enable Multi-Factor Authentication (MFA).",
    "Review firewall and security logs.",
    "Perform a complete antivirus scan.",
    "Preserve evidence for legal proceedings."
]

for i, rec in enumerate(recommendations, start=1):
    print(f"{i}. {rec}")

print("\n7. FINAL STATUS")
print("-" * 50)
print("Risk Level :", findings["Risk Level"])
print("Case Status: Investigation Completed")

print("\n" + "=" * 70)
print("      END OF DIGITAL FORENSIC INVESTIGATION REPORT")
print("=" * 70)
```
